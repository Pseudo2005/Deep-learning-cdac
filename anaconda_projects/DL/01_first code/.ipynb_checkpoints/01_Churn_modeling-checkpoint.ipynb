```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder
import pickle
```

```python
df=pd.read_csv(r"E:\Nitin_Pr\anaconda_projects\DL\datasets\Churn_Modelling.csv")
```

```python
df.head()
```

```python
#preprocess the data
df=df.drop(['RowNumber','CustomerId','Surname'],axis=1)
df
```

```python
## Encode categorical variable
le_gender=LabelEncoder()
df['Gender']=le_gender.fit_transform(df['Gender'])
df
```

```python
ohe_geography=OneHotEncoder(sparse_output=False)
geo_encoder=ohe_geography.fit_transform(df[['Geography']])
geo_encoder
```

```python
ohe_columns=ohe_geography.get_feature_names_out(['Geography'])
df_ohe=pd.DataFrame(geo_encoder,columns=ohe_columns)
```

```python
#combine both data
df=pd.concat([df.drop('Geography',axis=1),df_ohe],axis=1)
df
```

```python
df.info()
```

```python
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(le_gender,file)
with open('onehot_encoder_geo.pkl','wb')as file:
    pickle.dump(ohe_geography,file)   
```

```python
x=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)
```

```python
#Scale the feature
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
import sys
print(sys.executable)
print(sys.version)
```

```python
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

#Build our ann model

```python
model=Sequential([
     Dense(64,activation='relu',input_shape=(x_train.shape[1],)),#first hidden layer connected with  input layer
     Dense(32,activation='relu'), ##output layer
     Dense(1,activation='sigmoid') #output layer    
])
```

```python
model.summary()
```

```python
opt=tf.keras.optimizers.Adam(learning_rate=.01)
```

```python
model.compile(optimizer=opt,loss="binary_crossentropy",metrics=['accuracy'])
```
